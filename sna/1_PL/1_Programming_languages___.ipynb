#### 문자열 인덱싱

#### 문자열 슬라이싱



- 문자열의 특정 범위를 선택할 때 사용하는 연산자
- start : 추출하고자 하는 문자열의 첫 캐릭터의 인덱스
- stop : 추출하고자 하는 문자열 마지막 캐릭터의 인덱스에 1을 더한 값
- step : 생략가능, 생략 시 디폴트 값인 1이 사용 

````python
# 슬라이싱
s = '안녕하세요'

# '안녕' 만 출력
print(s[0:2:1])
print(s[0:2]) # step 1이면 생략 가능
print(s[:2]) # 처음부터는 start 생략 가능

# '하세' 만 출력
print(s[2:4])
print(s[-3:4])
print(s[-3:-1])

# '하세요' 만 출력
print(s[2:5])
print(s[2:]) # end를 생략하면 끝까지 출력

# 거꾸로 출력
print(s[::-1])
````

````python
a = "Hello Python, 반가워요"

# 'Python' 만 출력
print(a[6:12])

# '반가워요' 만 출력
print(a[14:])

# 'Hello' 만 출력
print(a[0:5])

# 반대로 출력
print(a[::-1])
````

### [리스트]

- 리스트 : 여러 개의 정보를 저장할 수 있는 공간

~~~
- 순서가 있고 데이터를 추가, 수정, 삭제가 가능
- 인덱싱, 슬라이싱 이용할 수 있음
- +, *, in, not in, len() 사용할 수 있음
- 변경 가능한 자료구조 (mutable)

````python
# 빈 리스트의 생성
a = []
print(a, type(a))

b = list()
print(b, type(b))

c = [1, 2, 3, 4, 5, '가', 0.5, True, [10, 20, 30]] # 다양한 자료형 포함 가능
print(c, type(c))
````

### [튜플]

#### 튜플 정의 및 필요성

- 리스트처럼 여러 데이터를 저장할 수 있는 자료형, 리스트와 유사하지만 값을 바꿀 수 없음, Immutable 데이타 타입
- 즉, Tuple은 요소를 추가하거나 갱신, 삭제 할 수 없음, index, count  등 메서드만 사용가능
- 인덱싱, 슬라이싱, +, *, in, not in, len() 사용 가능
- 튜플은 소괄호 ()로 생성, 소괄호 생략가능, 각 요소는 , 로 구분
- 튜플은 함수의 리턴에 많이 사용(여러 개의 값을 리턴 가능)
 

 - 왜 튜플을 사용할까?
   - 더 적은 메모리 공간 사용
   - 읽기 전용이므로 데이터가 실수로 손상될 염려가 없음
   - 리스트보다 간단한 구조라서 성능이 빠름

````python
# 튜플의 생성
t1 = (1,2,3)
print(t1, type(t1))

t2 = 1, 2, 3 # () 생략 가능
print(t2, type(t2))
````

````python
t3 = (10)
print(t3, type(10))

t3 = (10,) # 항목이 하나라도 쉼표 생략하면 안됨. 생략하면 정수로 여김
print(t3, type(t3))

t3 = 10, # () 생략해도 쉼표는 꼭 있어야한다.
print(t3, type(t3))
````

#### break와 continue

* break : 반복문에서 조건에 관계 없이 반복문을 벗어날 수 있는 명령, 가장 가까운 반복문 나감
* continue : 반복문에서 continue 문 뒤의 문장 즉, 반복문의 나머지를 수행하지 않고 반복문의 처음으로 다시 돌아가서 계속 반복을 수행하는 명령

````python
# 1에서 10까지 반복하도록 만들어보자 

flag = True
count = 1

while flag :
    print(count)
    if count == 10 :
        flag = False
    count += 1
````

````python
# break를 사용하여 1에서 10까지 반복하도록 만들어보자

count = 1

while True :
    print(count)
    if count == 10 :
        break
    count += 1

print('프로그램 종료')
````

````python
# continue를 사용하여  10까지 숫자중 3의 배수를 빼고 모두 더한 결과를 출력해보자. 

sum = 0

for i in range(11) :
    if i % 3 == 0 : # 3의 배수
        continue
    print(i, end = ' ')
    sum += i
print('=>', sum)
````

* 정답이 아닌 경우 숫자를 다시 입력하도록 하고 정답이면 '정답입니다!' 를 출력하고 게임을 종료한다. 

```
1~50사이의 숫자를 입력하세요 : 25

#### 객체 지향의 특징

* 캡슐화 Encapsulation: 
```
객체 속에 속성 +  행위를 포함
데이터와 클래스 내부에서만 사용하는 속성, 함수의 접근 차단  정보 은닉을 포함

````python
# 클래스 생성
class Car :
    # 속성
    brand = ''
    model = ''
    color = ''

    #메서드 정의
    def turn_on(self) :
        print(self.model, '자동차의 시동을 겁니다.')
        
    def turn_off(self) :
        print(self.model, '자동차의시동을 끕니다. ')
        
    def drive(self) :
        print(self.model, '자동차 운전 중입니다.')
        
    def show_info(self) :
        print(f'{self.brand}, {self.color}, {self.model} 자동차 입니다.')
````

````python
# 객체 생성

# mycar 객체 생성
mycar = Car()

#속성 값 지정
mycar.brand = '현대'
mycar.color = 'black'
mycar.model = '제니시스'

# 속성 호출
print(mycar.color)

#메서드 호출
mycar.show_info()
mycar.turn_on()
mycar.drive()
mycar.turn_off()

# yourcar 객체 설정
yourcar = Car()

#속성 값 지정
yourcar.brand = '기아'
yourcar.color = 'white'
yourcar.model = '카니발'

# 속성 호출
print(yourcar.color)

#메서드 호출
yourcar.show_info()
yourcar.turn_on()
yourcar.drive()
yourcar.turn_off()
````

#### 파일 포인터의 동작 원리 

- 파일을 열면(open), 파이썬은 파일의 가장 첫 번째 글자(인덱스 0)에 가상의 커서(포인터)를 놓음  
- 읽기/쓰기 후 이동: read(5)를 하면 5글자를 읽고, 포인터는 자동으로 6번째 글자 앞으로 이동  
- 연속 동작: 다음에 다시 read()를 호출하면, 포인터가 멈춰있던 그 지점부터 이어서 읽기 시작  
- 끝 도달: 포인터가 파일의 맨 마지막(EOF, End Of File)에 도달하면, 더 이상 읽을 내용이 없으므로 빈 문자열('')을 반환

##### seek() 함수: 포인터 강제 이동시키기

f.seek(offset, whence)  

    - offset: 이동할 바이트(Byte) 수입니다.  
    - whence: 어디를 기준으로 이동할지 결정합니다 (기본값은 0)    
    
       > 0: 파일의 맨 처음부터 시작 (가장 많이 씀)  
       > 1: 현재 포인터 위치부터 시작  
       > 2: 파일의 맨 끝부터 시작
