# 동적 웹페이지란?

동적 웹 페이지는 사용자의 상호작용이나 JavaScript 실행에 따라 콘텐츠가 변경되는 웹 페이지를 말합니다. 예를 들어, 스크롤을 내리면 추가 콘텐츠가 로드 되거나, 버튼 클릭 시 새로운 정보가 표시되는 경우 등이 이에 해당합니다. 이러한 페이지는 단순한 HTTP 요청으로는 모든 데이터를 가져올 수 없기 때문에 셀레니움과 같은 도구를 사용하여 브라우저를 제어하고 필요한 데이터를 수집해야 합니다.

# 셀레니움(Selenium) 이란 ? 

- 셀레니움은 웹 브라우저를 사람이 사용하는 것처럼 똑같이 제어할 수 있게 해주는 자동화 도구
- 기존에 사용하신 방식(requests)으로 크롤링이 안 되는 동적 사이트 웹크롤링할 때 사용
    - 동적 페이지 수집: 인스타그램이나 유튜브처럼 스크롤을 내려야 데이터가 더 로드되는 사이트
    - 자바스크립트 실행: 페이지에 접속한 뒤 자바스크립트가 실행되어야 데이터가 나타나는 사이트
    - 사용자 상호작용: 아이디/비밀번호를 입력하고 '로그인' 버튼을 눌러야만 볼 수 있는 페이지
    - 클릭 및 팝업: "더보기" 버튼을 눌러야 기사가 다 나오거나, 팝업창을 닫아야 하는 경우

# 셀리니움 작동 원리

- 셀레니움은 파이썬 코드와 브라우저 사이에서 **'브라우저 드라이버(WebDriver)'**라는 중개인을 두고 소통
- 예
    - 파이썬 코드: "네이버 접속해서 검색창에 '환율'이라고 쳐줘."
    - WebDriver: 파이썬의 명령을 브라우저가 알아들을 수 있게 전달
    - 브라우저: 실제로 검색창을 찾아서 글자를 입력하고 엔터를 침

## 웹드라이버 란?

- 셀레니움으로 웹브라우저를 조작하려면 웹드라이버 필요
- 비유하자면 웹브라우저는 자동차, 웹드라이버는 자동차를 제어하는 모듈(핸들, 패달, 계기판 등), 셀레니움은 운전사
- 크롬, 파이어폭스, 사파리 등의 웹브라우저는 각기 다른 웹드라이버를 가지고 있으며 가각 브라우저 제작사에서 만들어 제공

### 웹드라이버 설치 방법

- 웹드라이버 매니저를 사용하여 자동으로 웹드라이버를 설치하고 관리해주는 방법
- <b>셀레니움 매니저를 사용하여 자동으로 웹드라이버를 관리, 셀레니움 매니저는 셀리니움 버전 4.6 부터 포함(별도 설치 필요 없음)</b>
- 수동 설치: 웹드라이버 파일을 수동으로 다운로드해서 설치하여 사용

## 드래그 앤 드롭하기

- 드래그 앤 드롭을 실행하기 위해 ActionChains 클래스 사용
- ActionChains 클래스
    - 셀레니움에서 마우스 클릭, 드래그 앤 드롭, 키보드 입력과 같은 복잡한 사용자 동작을 시뮬레이션할 때 사용하는 클래스로 마우스와 키보드 동작을 연속으로 실행할 수 있음
    - 동작을 즉시 실행하지 않습니다. 대신 호출하는 명령들을 차곡차곡 대기열에 저장해 두었다가, 마지막에 .perform()을 호출하는 순간 한꺼번에 순차적으로 실행

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.action_chains import ActionChains
import time

driver = webdriver.Chrome()
driver.get('http://www.naver.com')

driver.implicitly_wait(10)
# 드래그할 대상 요소와 드롭할 목표 요소를 지정  
# 스토어 #shortcutArea > ul > li:nth-child(4) > a > span.service_icon.type_shopping
source = driver.find_element(By.CSS_SELECTOR,'#shortcutArea > ul > li:nth-child(4) > a > span.service_icon.type_shopping')
# 검색창 id query
target = driver.find_element(By.CSS_SELECTOR, '#query')
# 드래그 앤 드롭 실행
actions = ActionChains(driver)
actions.drag_and_drop(source, target).perform()

time.sleep(3)
# [Enter] 사용하기
target.send_keys(Keys.ENTER)
```

## 특정 요소 위치로 스크롤바 이동하기

```python
from selenium import webdriver

driver = webdriver.Chrome()
driver.get('https://www.naver.com')

# 페이지에서 [증시] 코너의 [증시] 요소 찾기, 링크 텍스트에 어떤 키워드가 있을 때 사용
driver.implicitly_wait(10)
element = driver.find_element(By.PARTIAL_LINK_TEXT, '증시')

# 해당 요소가 화면 중앙으로 오도록 부드럽게 스크롤바를 조정
#scrollIntoView() 특정요소가 보이도록 스크롤바 이동시키는 javascript 함수
# behavior 스크롤 동작,block 수직방향, inline 수평방향 
driver.execute_script("arguments[0].scrollIntoView({behavior:'smooth', block:'center',inline:'center'})",element)
```

## 브라우저 창 전환하기

## 개별 데이터 수집하기

```python
from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
import time

# 크롬 드라이버 객체 생성 및 지정된 URL로 이동
driver = webdriver.Chrome()
driver.get('https://land.naver.com/news/region.naver') #네이버페이 부동산 - 뉴스 - 우리동네뉴스

# "지역선택" 요소 찾기
xpath = '//*[@id="searchForm"]/dl/dt/span'
element = driver.find_element(By.XPATH, xpath)

# 시/도 드롭다운 요소를 찾아서 클릭
city_list = '//*[@id="city_no"]'
driver.find_element(By.XPATH, city_list).click()
# 각 시/도별 요소 찾기
city = '인천'  # 시/도를 지정

# 지정한 시/도에 해당하는 요소의 XPath를 동적으로 생성
city_element = f'//select[@id="city_no"]/option[contains(text(), "{city}")]'

# 해당 요소가 클릭 가능할 때까지 최대 10초간 대기
WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, city_element)))
driver.find_element(By.XPATH, city_element).click()  # 지정한 시/도의 요소를 찾아 클릭

# 시/군/구 지정
town = '부평구'
# 시/군/구 드롭다운 영역 다루기(시/군/구 값이 주어진 경우)
if town:
 town_list = '//*[@id="dvsn_no"]'
 driver.find_element(By.XPATH, town_list).click()
 # 지정한 시/군/구에 해당하는 요소의 XPath를 동적으로 생성
 town_element = f'//*[@id="dvsn_no"]/option[contains(text(), "{town}")]'
 # 해당 요소가 클릭 가능할 때까지 최대 10초간 대기
 WebDriverWait(driver, 10).until(EC.element_to_be_clickable((By.XPATH, town_element)))
 driver.find_element(By.XPATH, town_element).click() # 지정한 시/군/구 요소를 찾아 클릭

time.sleep(2) # 지역 선택 후 뉴스가 로딩될 때까지 충분히 대기 (매우 중요!)

# ➊ 전체 뉴스 데이터 수집 (반드시 find_elements 복수형 사용)
# li 태그까지 경로를 지정해야 하나씩 꺼내올 수 있습니다.
news_items = driver.find_elements(By.XPATH, '//*[@id="content"]/div[2]/ul/li')

rows = [] # ➋ 빈 리스트 생성

for data in news_items:
    try:
        # ➌ 뉴스 링크 추출
        # 경로를 './/dt//a'로 수정하여 이미지가 있든 없든 제목 링크를 찾도록 합니다.
        #./ 자식, .// 후손
        link_element = data.find_element(By.XPATH, './/dt//a')
        link = link_element.get_attribute('href')
        
        # ➍ 제목과 내용 추출
        # 텍스트가 없는 빈 항목이나 광고 요소가 섞일 경우를 대비해 예외처리를 합니다.
        text_data = data.text.strip().split('\n', maxsplit=1) # maxsplit=1처음 나타나는 \n 한번만 분리 
        
        # if len(text_data) >= 2:
        title = text_data[0]
        contents = text_data[1]
        # else:
        #     title = text_data[0]
        #     contents = "" # 내용이 없는 경우 빈 문자열 처리
        
        # ➎ row 생성
        row = {
            '뉴스기사명': title,
            '내용': contents,
            '링크': link
        }
        rows.append(row)
        
    except Exception as e:
        # 에러가 난 항목은 건너뛰고 다음 뉴스를 수집합니다.
        continue
    
# 결과 출력
if rows:
    print('수집된 뉴스 개수:', len(rows))
    print('첫 번째 행 출력: ', rows[0])
else:
    print("데이터를 찾지 못했습니다. XPath를 확인해 보세요.")
```
