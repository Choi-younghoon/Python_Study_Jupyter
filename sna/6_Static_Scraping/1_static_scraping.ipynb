# 웹스크레이핑 이해하기

## 웹스크레이핑 이란?

- 인터넷에서 데이터를 추출하는 과정

## 웹스크레이핑 주의사항


### 웹사이트의 이용 약관 준수

- 데이터를 스크레이핑 할 웹사이트의 이용 약관을 확인해보고 데이터 수집을 명시적으로 금지하는 웹사이트에서 웹스크래핑을 시도하면 안됩니다.
  

### 개인정보 수집 금지

- 스크레이핑 안 데이터에 개인 정보가 포함되 있다면 정보를 보호하기 위한 적절한 조치를 취해야합니다.
- 데이터를 안전하게 저장하고, 불필요한 정보는 삭제하여 개인 정보 보호 법뮬을 준수해야 합니다.

### robots.txt 파일 확인

- 웹사이트 url에 '/robots.txt'를 붙여 입력하면  robots.txt 파일 확인할 수 잇습니다.
- 이 내용은 웹사이트에서 데이터 수집을 허용하는 범위나 지침에 대한 것입니다.

### 잦은 스크레이핑 요청 금지

- 단기간에 너무 많은 스크레이핑 요청을 보내면 웹사이트 서버에 부하를 줄 수 잇습니다.
- 자주 요청을 보내면 요청을 보내는 컴퓨터나 IP 주소등을 차단할 수 있으므로 주의하기 바랍니다.
- 보통 웹스크레이핑에 요청 지연 시간을 설정하여 방지합니다.

### 시간 지남에 따른 정보 변경 시 업데이트

- 웹사이트는 시간이 지나면서 주소인 URL 변화, 레이아웃과 구조가 변경되는 HTML 변화가 있을 수 있습니다.
- 웹스크레이핑을 유지 관리하고 변경사항을 수용하도록 업데이터 해야합니다.

## 웹크롤링과 웹스크레이핑 차이

- **웹크롤링(Crawling)은 웹사이트를 순회하면서 대규모의 데이터를 수집하는 기술**
    - 웹크롤링하는 프로그램을 웹크롤러 또는 스파이더, 로봇 등으로 부릅니다.
    - 웹사이트의 전체구조를 따라 이동하면서 웹페이지간의 관계를 파악하고 웹페이지의 내용을 수집한 다음 검색엔젠의 데이터베이스에 저장, 저장된 데이터를 가지고 검색엔진이 검색결과 제공
- **웹스크레이핑(Scraping) 특정 웹페이지에서 필요한 데이터를 추출하는 과정**
    - 웹크롤링보다 좁은 범위에서 특정 데이터만을 추출하는 기술을 의미


## 정적 웹페이지와 동적 웹페이지

- 정적 웹페이지: 언제 접속해도, 누가 접속해도 똑같은 내용을 보여주는 페이지
    - 서버에 이미 저장된 HTML 파일 그대로를 사용자에게 전달하는 방식
- 동적 웹페이지: 상황에 따라 사용자 요청에 따라 내용이 실시간으로 변하는 페이지
    - 사용자가 누구인지, 어떤 검색어를 입력했는지 등에 따라 서버가 데이터베이스에서 정보를 찾아 즉석해서 페이지를 구성해 보여주는 방식

## 웹페이지의 구성

- **HTML (HyperText Markup Language):** 웹페이지의 **구조와 내용**을 담당합니다. 하이퍼링크를 통해 문서 간의 연결을 정의하고, 텍스트·이미지 등 웹의 뼈대를 만듭니다.
- **CSS (Cascading Style Sheets):** 웹페이지의 **시각적 디자인**을 담당합니다. 레이아웃, 색상, 글꼴 등을 설정하여 웹 문서를 보기 좋게 꾸며줍니다.
- **JavaScript:** 웹페이지의 **동적 처리와 상호작용**을 담당합니다. 사용자 클릭에 반응하거나 데이터를 실시간으로 갱신하는 등 생동감 있는 기능을 구현합니다.

### HTML 문서 구조

- HTML 문서는 웹페이지의 구조를 정의하는 마크업 언어 이며, 태그, 요소, 속성으로 구성됨.

- **태그(Tag)**:웹페이지의 요소를 정의하는 단위. 시작 태그(`<p>`)와 종료 태그(`</p>`)가 있음, 요소는 부모와 자식의 관계가 있음
- **요소(Element)**: 태그로 감싸진 것, 시작 태그 + 내용 + 종료 태그로 구성됨, 웹페이지에서 특정 역할을 수행
- **속성(Attribute)**: 태그에 **추가 정보**를 부여하는 것

### CSS 구조

- CSS(Cascading Style Sheets)**는 HTML 문서에 **디자인(스타일)**을 적용하기 위한 **스타일 정의 언어**
- HTML이 웹 페이지의 **구조**를 정의한다면, CSS는 그 구조에 **색, 크기, 배경, 정렬 등 시각적스타일**을 입힘

## 서버에 요청과 응답을 실행하는 Requests 라이브러리

- 웹 서버에 HTTP(HyperText Transfer Protocol) 요청을 보내고 응답(HTML, JSON 등)을 받아오는 가장 표준적인 라이브러리

- HTTP 요청 방식

    - <b>GET(조회): 서버로 부터 특정 데이터를 가져올 때 사용</b>
    - POST(등록): 서버에 데이터를 전달하여 새로운 리소스를 생성하거나 처리를 요청할 때 사용
    - PUT(수정): 서버에 있는 리소스를 전체적으로 업데이터 할 때 사용
    - DELETE(삭제): 서버의 특정 리소스를 삭제할 때 사용

- response에 들어 있는 값   
      - status_code: 요청의 성공 또는 실패 여부를 나타내는 응답코드  
      - text: 웹페이지의 HTML 문서  
      - headers: 콘텐츠 유형, 인코딩, 서버 정보 등이 있는 헤더  
      - cookies: 쿠키 정보  
      - elapsed: 서버가 요청을 받고 응답을 완료하는데 걸린 시간

## 유저 에이전트 이해하기

### User-Agent 란?

- <b>User-Agent(사용자 에이전트)</b>는 웹 브라우저나 앱이 서버에 접속할 때 "나는 어떤 기기이고, 어떤 브라우저를 쓰고 있어"라고 자신을 소개하는 일종의 자기소개서와 같은 문자열
- HTTP 요청 헤더(Request Header)에 포함되어 서버로 전달

### User-Agent 역할

- 기기별 최적화: 접속자가 스마트폰(모바일)인지 PC(데스크톱)인지 판단하여 모바일용 웹페이지를 보여줄지 결정합니다.

- 브라우저 호환성: 사용 중인 브라우저(Chrome, Safari, Edge 등) 버전에 맞춰 최적화된 콘텐츠를 제공합니다.

- 통계 수집: 웹사이트 방문자들이 어떤 브라우저나 OS를 주로 사용하는지 분석하는 용도로 쓰입니다.

- <b>보안 및 차단: 특정 브라우저나 비정상적인 접근(웹 크롤링 봇 등)을 식별하여 차단하거나 허용합니다.</b>

<font color='red'>파이썬의 requests 라이브러리를 사용하여 웹 데이터를 수집할 때, 기본 User-Agent는 보통 python-requests/2.x.x로 설정되므로, 많은 웹사이트들이 "사람이 아닌 봇(Bot)"의 접근을 막기 위해 기본 User-Agent 값을 가진 요청을 차단하곤 합니다. 이럴 때 마치 브라우저에서 접속한 것처럼 User-Agent를 조작(Spoofing)해야 합니다.</font>

직접 접속할때는 데이터가 보이는데 웹스크레이핑 프로그램으로 같은 url 요청할 때 오류 메시지가 나오면 User-Agnet 조작

### User-Agent 확인하기

- 구글에 User Agent String 으로 검색

Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36

<span style="color: blue;"><u>**유저 에이전트 조작하기**</u></span>

```python
import requests

url = 'http://www.google.com'
# 헤더에 유저 에이전트 값 지정
headers = {'User-Agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36'}

# headers 옵션을 설정하여 get()함수 실행
response = requests.get(url, headers=headers)
response.raise_for_status()
print(response.status_code)
```

## 실습 html 파일 만들기

```python
html = '''
<html>
<head>
  <title>온라인 쇼핑몰</title>
</head>
<body>
  <h1> 장바구니
    <p id='clothes' class='name' title='라운드티'> 라운드티
      <span class = 'number'> 25 </span>
      <span class = 'price'> 29000 </span>
      <span class = 'menu'> 의류</span>
      <a href = 'http://www.naver.com'> 바로가기 </a>
    </p>
    <p id='watch' class='name' title='시계'> 시계
      <span class = 'number'> 28 </span>
      <span class = 'price'> 32000 </span>
      <span class = 'menu'> 액세서리 </span>
      <a href = 'http://www.facebook.com'> 바로가기 </a>
    </p>
  </h1>
</body>
</html>
'''
```

## find( ), find_all( ) 함수로 요소 찾기

- soup.find(): 파싱한 HTML 문서에서 조건에 부합하는 첫 번째 요소를 찾음
- soup.find_all(): 파싱한 HTML 문서에서 조건에 부합하는 모든 요소를 찾아 파이썬 리스트로 반환

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, 'html.parser')
print(soup.find_all('p'))
```

```python
soup = BeautifulSoup(html, 'html.parser')
print(soup.find('p'))
```

- html.parser HTML 코드를 해석해서 구조(Tree)를 만들어주는 '번역기

## select_one( ), select( ) 함수로 요소 찾기

- soup.select(): 파싱한 HTML 문서에서 CSS 선택자 조건에 부합하는 모든 요소를 찾아 파이썬 리스트로 반환

## [실습] 명언사이트에서 명언과 작가 추출정보를 추출하기

-  https://quotes.toscrape.com 사이트에서 **명언(text)**과 작가(author) 정보를 크롤링해서 출력하기

```python
import requests
from bs4 import BeautifulSoup

# 1. 요청할 웹페이지 주소
url = "https://quotes.toscrape.com"
headers = {'User-Agent': 'Mozilla/5.0'}

# 2. 웹페이지 요청 & 응답 받기
res = requests.get(url, headers=headers)

# 3. HTML 파싱하기
soup = BeautifulSoup(res.text, 'html.parser')

# 4. 명언이 들어 있는 박스들 찾기
quotes = soup.find_all('div', class_ = 'quote')
# print(quotes)

# 5. 각 명언과 작가 출력
for q in quotes :
    text = q.find('span', class_ = 'text').text
    author = q.find('small', class_ = 'author').text
    print(f'{text} - {author}')
```

## [실습]멜론 주간 차트 1 ~100 위까지 순위, 노래제목, 가수 추출해서 출력하기

- https://www.melon.com/chart/week/index.htm- 

```python
import requests
from bs4 import BeautifulSoup

## 멜론 주간 차트 1 ~ 100위 크롤링하기 ##

#크롤링을 방지하는 사이트에서 만약 User-Agent 를 점검하는 로직이 있다면,
# header에 유저정보를 넣어주면 크롤링 됨
header = {'User-Agent': 'Mozilla/5.0'}

#멜론 URL 주소의 웹페이지 내용을 가져온다.
res = requests.get('https://www.melon.com/chart/week/index.htm', headers = header)

#웹페이지의 데이터를 파이썬이 읽기 쉽게 분해(파싱)하여 객체로 만듬
parse = BeautifulSoup(res.text, 'html.parser')

#해당 속성의 값을 가지는 태그를 검색
#div태그의 class 속성이 ellipsis rank01, ellipsis rank02, ellipsis rank03인 값을 찾아옴
#여러 개를 찾을 때는 findAll
titles = parse.findAll('div', {'class': 'ellipsis rank01'})
singers = parse.findAll('div', {'class': 'ellipsis rank02'})


for i in range(100):
    rank = i + 1
    
    # 제목 추출: 첫 번째 a 태그의 텍스트
    title = titles[i].find('a').text
    
    # 가수 추출: 첫 번째 a 태그의 텍스트 (여러 명일 경우 첫 번째 가수만 출력)
    singer = singers[i].find('a').text
    
    # 출력부 수정: 앨범(album) 변수 삭제
    print(f"{rank}, {title}, {singer}")
```

```python
import requests
from bs4 import BeautifulSoup


# ➊ 페이지 수, 카테고리, 날짜를 전달받아 뉴스 기사의 URL을 추출하는 함수 make_urllist( ) 정의
def make_urllist(page_num, category1, category2, date):
    urllist = []
    for i in range(1, page_num + 1) :
        url = 'https://news.nate.com/subsection?cate='+category1+'&mid=n'+category2+'&type=c&date='+str(date)+'?page='+str(i)
        news = requests.get(url)
        soup = BeautifulSoup(news.text, 'html.parser')
        news_list = soup.select('.postSubjectContent .mduSubjectList .mlt01 a')
        for line in news_list :
            urllist.append(line.get('href'))

    return urllist
```
